## Ejemplo propuesto 3: sensibilidad de los estados y controlabilidad

**Objetivo** ¿Cómo responde el sistema a perturbaciones de sus variables de estado? ¿Puede modificarse su comportamiento mediante determinadas entradas o acciones?
Aquí es importante precisar el término: estudiar perturbaciones de estados no equivale necesariamente a analizar la controlabilidad en sentido matemático.
