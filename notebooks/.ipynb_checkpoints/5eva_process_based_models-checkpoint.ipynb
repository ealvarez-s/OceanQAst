# Modelo Basado en Procesos + integración de datos 
## La propuesta: Modelo 0D con integración de clorofila-a de satélite
**Ventajas** Un modelo basado en procesos respeta las leyes biológicas y físicas; la calibración corrige parámetros y la asimilación el estado con datos reales.

**Desventajas** Requiere calibrar muchos parámetros biológicos difíciles de medir.

**Objetivo** Usar las observaciones reales de clorofila-a del satélite (como MODIS o Sentinel) para "corregir" las desviaciones del modelo biológico en tiempo real (por ejemplo, mediante un Filtro de Kalman). Si el modelo simula que el fitoplancton va a caer pero el satélite muestra que sigue alto, la asimilación ajusta las ecuaciones del modelo para que la predicción del día siguiente sea mucho más realista y físicamente coherente.
