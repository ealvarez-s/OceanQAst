# Machine Learning: Random Forest
## La propuesta: Random Forest (RF)
**Ventajas** El océano real es altamente no lineal y las variables interactúan de formas complejas. Random Forest no requiere asumir distribuciones estadísticas estrictas. Encuentra patrones complejos e interacciones entre muchas variables.

**Desventajas** Es una "caja negra"; no respeta las leyes de la física por sí solo.

**Objetivo** Predecir la presencia de una especie marina (o un bloom de algas nocivas) a partir de múltiples variables ambientales simultáneas: salinidad, profundidad, oxígeno disuelto, corrientes y temperatura. El algoritmo aprende los umbrales óptimos combinando estas variables sin que tú tengas que definir la ecuación matemática.
