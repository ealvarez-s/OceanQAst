# Sensibilidad a perturbaciones

1. Problema científico: ¿Qué pregunta oceanográfica puede abordar?
El análisis de sensibilidad permite estudiar cómo los cambios en los parámetros de un modelo afectan a sus resultados. En oceanografía, resulta útil para identificar qué parámetros condicionan más la evolución simulada de variables como la clorofila, los nutrientes o la biomasa fitoplanctónica.

Este análisis ayuda a comprender el comportamiento del modelo, identificar parámetros que merecen mayor atención y orientar futuros procesos de calibración. No determina por sí solo qué valores de los parámetros son correctos ni demuestra que el modelo represente adecuadamente el sistema real.

4. Validación: ¿Cómo sabemos si predice razonablemente bien?
La interpretación de los resultados depende de los parámetros seleccionados, sus rangos de variación y las salidas analizadas. Por tanto, los índices de sensibilidad describen la influencia de los parámetros dentro del experimento definido, pero no indican por sí solos si sus valores son correctos ni si el modelo representa adecuadamente el sistema real.

5. Limitaciones: ¿Cuándo falla y qué conclusiones no permite extraer?
Los resultados dependen de la selección de parámetros, sus rangos, las variables de salida y las condiciones fijadas. El análisis puede ser costoso si el modelo requiere mucho tiempo de ejecución, y los índices pueden ser poco estables con un número insuficiente de muestras.

Además, una sensibilidad baja dentro del rango explorado no implica que un parámetro sea irrelevante bajo otras condiciones. El análisis tampoco sustituye la validación del modelo frente a observaciones ni la calibración de sus parámetros.
