La evaluación predictiva consiste en comparar las predicciones con las observaciones correspondientes mediante métricas que describan diferentes aspectos del error y del comportamiento del modelo.

Siguiendo el enfoque de evaluación de modelos marinos propuesto por Stow et al. (2009), se pueden considerar varias métricas complementarias:

Error medio (AE): permite identificar el sesgo medio de las predicciones. Su signo indica si el modelo tiende a sobreestimar o subestimar, según la convención adoptada.
Error absoluto medio (AAE): resume la magnitud media de los errores sin que los signos se compensen.
Raíz del error cuadrático medio (RMSE): penaliza especialmente los errores grandes.
Eficiencia del modelo (MEF): compara el error del modelo con la variabilidad de las observaciones y permite valorar su rendimiento frente a una referencia basada en la media observada.
Correlación: evalúa hasta qué punto las predicciones reproducen la variabilidad conjunta de las observaciones, aunque una correlación elevada no garantiza que los valores predichos sean precisos.
Índice de fiabilidad (RI): evalúa las discrepancias relativas entre observaciones y predicciones mediante una formulación logarítmica, por lo que requiere datos positivos y un tratamiento cuidadoso de los valores nulos o negativos.
