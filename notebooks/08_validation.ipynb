## Comparar los tres enfoques
**Objetivo** Evaluar skill en varios horizonte de predicción. Investigar no solo qué modelo predice mejor, sino también qué información aporta cada enfoque.
- ETS/SARIMAX: referencias estadísticas con las que evaluar la capacidad predictiva.
- Random Forest: referencia no lineal basada en las relaciones aprendidas de los datos.
- Modelo basado en procesos: predicción fundamentada en la dinámica biogeoquímica.
- Modelo basado en procesos con asimilación: evaluar si incorporar las observaciones recientes mejora el pronóstico frente a ejecutar el modelo sin asimilación.
  
## Ejemplo propuesto: Stow et al. 2009
La comparación debe hacerse con los mismos periodos de evaluación y horizontes de predicción. En particular, la asimilación solo puede utilizar observaciones disponibles hasta el instante de emisión de cada pronóstico; no debe acceder a datos futuros del periodo evaluado.

Métricas comunes, validación temporal, incertidumbre, evaluación de episodios extremos y comparación de indicadores derivados de las observaciones y las simulaciones.
