# Integración de datos
Calibración como una forma de integrar información observacional en la configuración de un modelo, ajustando sus parámetros para mejorar el acuerdo entre las simulaciones y las observaciones.

1. Problema científico: ¿Qué pregunta oceanográfica puede abordar?
Los modelos oceanográficos representan procesos físicos, químicos y biológicos mediante ecuaciones y parámetros que, en ocasiones, presentan incertidumbre o no pueden determinarse directamente a partir de observaciones.

La calibración consiste en ajustar determinados parámetros del modelo para mejorar el acuerdo entre sus resultados y los datos observados. En oceanografía, este procedimiento puede ayudar a evaluar configuraciones de modelos biogeoquímicos, estimar parámetros difíciles de medir directamente y analizar qué combinaciones de parámetros reproducen mejor la dinámica observada.

2. Datos necesarios: ¿Qué variables, resolución temporal y longitud de serie requiere?
**Modelo ejecutable**: una configuración que pueda simularse repetidamente con diferentes valores de los parámetros.
**Configuración de referencia**: condiciones iniciales, forzamientos y demás parámetros que se mantendrán fijos, salvo que se decida calibrarlos también.
**Observaciones**: medidas de variables comparables con las salidas del modelo, con unidades, escalas temporales y profundidades compatibles.
**Parámetros ajustables**: parámetros seleccionados para la calibración y sus rangos plausibles.
**Criterio de ajuste**: una función objetivo que cuantifique la discrepancia entre las observaciones y las simulaciones.

3. Fundamento: ¿Qué variables, resolución temporal y longitud de serie requiere?
La calibración puede formularse como un problema de optimización en el que se buscan los valores de los parámetros que minimizan una función objetivo que cuantifica la discrepancia entre las observaciones y las simulaciones.

La solución obtenida depende de la función objetivo, los rangos de los parámetros y la configuración del algoritmo. Por ello, el resultado debe evaluarse científicamente y no interpretarse automáticamente como el conjunto de parámetros verdadero.

5. Limitaciones: ¿Cuándo falla y qué conclusiones no permite extraer?
La calibración depende de la calidad y representatividad de las observaciones, de la función objetivo, de los rangos de los parámetros y del algoritmo de optimización empleado.

Aunque Differential Evolution se considera un método de optimización global, eso no significa que garantice encontrar el mínimo global. Si el modelo es costoso, el presupuesto de simulaciones puede limitar la exploración del espacio de parámetros.

Además, la incertidumbre estructural del modelo no desaparece al ajustar sus parámetros. Si faltan procesos relevantes o las ecuaciones no representan adecuadamente el sistema, la calibración puede compensar esos errores mediante valores de parámetros poco realistas.

También puede haber problemas de identificabilidad: varios conjuntos de parámetros pueden generar simulaciones parecidas. Por tanto, el resultado debe interpretarse como una solución condicionada por los datos, las hipótesis y el procedimiento utilizado, no necesariamente como una estimación única de los parámetros verdaderos.

## A mayores: asimilación de datos
Una posible ampliación de este ejemplo es la asimilación de datos, un conjunto de métodos que combina las observaciones con la información proporcionada por un modelo para mejorar la estimación del estado de un sistema oceanográfico.

A diferencia de una calibración convencional, que suele buscar valores de parámetros que mejoren el ajuste entre simulaciones y observaciones, muchos métodos de asimilación actualizan las estimaciones del estado del sistema a medida que se incorporan observaciones. Esto permite explorar aplicaciones como la estimación de perfiles de variables oceanográficas o la mejora de las condiciones iniciales de una simulación.

La asimilación requiere considerar aspectos adicionales, como la incertidumbre del modelo y de las observaciones, el método de actualización y la frecuencia con la que se incorporan los datos. Su implementación queda fuera del alcance de este notebook, pero constituye una posible línea de desarrollo futuro para OceanQAst.
